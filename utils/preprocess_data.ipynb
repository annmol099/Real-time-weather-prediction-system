```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
```

```python
# Raw Data Load Karo
df = pd.read_csv("../data/raw_data.csv")
print(f"Raw data loaded: {df.shape}")
print(df.head())
```

```python
# Table Format Mein Dikhao
from IPython.display import display

# Pehli 10 rows table mein
display(df.head(5))

print(f"\nTotal Rows: {len(df)}")
print(f"Total Columns: {len(df.columns)}")
```

```python
# Missing Values 
print("Missing Values:")
print(df.isnull().sum())
```

```python
print(f"\nData Types:")
print(df.dtypes)
```

```python
# Time column String se Datetime mein convert karo
df['time'] = pd.to_datetime(df['time'])
df['time']
```

```python
df.head()
```

```python
# Time Features 
df['hour'] = df['time'].dt.hour
df['day'] = df['time'].dt.day
df['month'] = df['time'].dt.month
df['dayofweek'] = df['time'].dt.dayofweek

print("Time features ban gaye!")
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['city_encoded'] = le.fit_transform(df['city'])

print("Data ready hai!")
df.head()
```

```python
for i in range(1, 5):
    df[f'temp_day{i}'] = df.groupby('city')['temperature_2m'].shift(-24*i)
    df[f'humidity_day{i}'] = df.groupby('city')['relative_humidity_2m'].shift(-24*i)
    df[f'wind_day{i}'] = df.groupby('city')['windspeed_10m'].shift(-24*i)
    df[f'rain_day{i}'] = df.groupby('city')['precipitation'].shift(-24*i)
```

##### 🧹 Remove Missing Values

We remove rows containing NaN values to ensure clean and reliable data for further analysis and modeling.

```python
df = df.dropna()

print(f"Data ready: {df.shape}")
print(df.columns.tolist())
```

```python
import joblib

# Label Encoder
joblib.dump(le, "../model/label_encoder.pkl")

# saved in CSV
df.to_csv("../data/processed_data.csv", index=False)
print("processed_data.csv saved! ✅")
```

```python
# Column Names Rename
df = df.rename(columns={
    'temperature_2m': 'temp',
    'relative_humidity_2m': 'humidity',
    'windspeed_10m': 'wind',
    'pressure_msl': 'pressure',
    'precipitation': 'rain',
    'cloudcover': 'cloud',
    'city': 'city',
    'latitude': 'lat',
    'longitude': 'lon',
    'hour': 'hour',
    'day': 'day',
    'month': 'month',
    'dayofweek': 'weekday',
    'city_encoded': 'city_id'
})

print("Columns rename ho gaye!")
print(list(df.columns))
```

```python
# Processed Data Save Kiya
df.to_csv("../data/processed_data.csv", index=False)

print("processed_data.csv save is done!")
print(f"Total rows: {len(df)}")
print(f"Total columns: {len(df.columns)}")
print(f"\nColumns: {list(df.columns)}")
```
